# Imports

```python
import pandas as pd
```

# Getting DF

```python
df = pd.read_csv("final_df.csv")
df
```

# Cleaning columns

```python
columns_to_drop = ["ip_addr",
    "spotify_track_uri",
    "episode_name",
    "episode_show_name",
    "spotify_episode_uri",
    "audiobook_title",
    "audiobook_uri",
    "audiobook_chapter_uri",
    "audiobook_chapter_title",
    "offline_timestamp",
    "in_playlist",
    "lat_long",
    "track_id",
    "ms_played",
    "duration_ms",
    "Unnamed: 0"
    ]

for i in columns_to_drop:
    try:
        df = df.drop(columns=i)
    except:
        print(f"A coluna {i} não existe no df")
df.columns
```

# Export Df

```python
df.to_csv("clean_df.csv", index=False)
```
